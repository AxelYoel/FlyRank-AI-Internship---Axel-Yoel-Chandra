## 2. The question: decision, action, cost of a wrong call

My question: which observable content and search signals (position,
freshness, word count, engagement, CTR, etc.) are actually associated with
a page's visibility, clicks, or engagement — and how strong is each
association?

**Decision this improves:** right now, a content or SEO team has to guess
or rely on general SEO folklore when deciding what to watch or prioritize
across a large content inventory. This project replaces that guesswork
with evidence about which signals actually move with performance in
FlyRank's own data.

**Who acts on it:** a content strategist or SEO analyst, who uses the
signal report to decide what to monitor and where to focus review effort
— for example, whether freshness matters more than word count, or whether
position decay tends to show up before click declines.

**Cost of a wrong call:** since this is a signal-discovery project rather
than a direct action queue, the main risk is misdirected attention rather
than a bad one-off decision. If I claim a signal is meaningfully
associated with performance when it isn't (or miss one that actually
matters), teams could spend time optimizing or tracking the wrong thing —
for instance, obsessing over word count while a stronger signal like
position or freshness goes unnoticed. Because this is observational, not
causal, the bigger risk is someone treating a correlation as a guaranteed
lever to pull, rather than a pattern worth investigating further.

## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

```python
data['ctr'].describe()
```

```python
#correlation between position and CTR
print(data[["avg_position", "ctr"]].corr())
```

The raw correlation between `avg_position` and `ctr` is weak (r = -0.073)

```python
#correlation between word count and impressions
print(data[["word_count", "impressions_90d"]].corr())
```

`word_count` shows only a weak positive correlation with `impressions_90d`
  (r = 0.163), suggesting content length alone is a modest signal at best,
  not a strong lever

```python
#engagement by trend direction
print(data.groupby("trend_direction")["engagement_rate"].mean())
```

Engagement rate does not move in a simple pattern by `trend_direction`:
  "down" pages average 2.44 vs "flat" pages at 1.37. it counters to the
  assumption that declining pages always show worse engagement

```python
#CTR by position tier
print(data.groupby("position_tier")["ctr"].mean().sort_values(ascending=False))
```

grouping CTR by `position_tier` reveals a much clearer pattern:
  mean CTR drops steadily from top_3 (1.48) to page_1 (0.65) to deep (0.15).
  This suggests the position-CTR relationship is not well captured by a
  linear correlation and may need a non-linear or grouped approach

## 4. Careful words: what I can and can't claim

**What I can say:**
- This work is *observational*, not experimental. I can describe
  associations between signals (like position, word count, freshness) and
  outcomes (like impressions, CTR, engagement) — not prove that one causes
  the other.
- I can say a signal is "associated with," "correlated with," or "linked
  to" an outcome, and describe how strong that relationship appears to be
  (e.g. weak, moderate) based on correlation coefficients or grouped
  comparisons.
- I can flag patterns that are counterintuitive or worth further
  investigation (e.g. declining pages showing higher engagement than flat
  pages) without claiming to explain why they happen.
- I can say this analysis is decision-support: it points a content
  strategist toward signals worth watching, not a guaranteed lever to
  pull.

**What I cannot say:**
- I cannot claim that changing a signal (e.g. increasing word count) will
  *cause* a page's performance to improve — that would require an
  experiment or causal design, which this data doesn't provide.
- I cannot claim to have reverse-engineered or predicted anything about
  Google's actual ranking algorithm. All I have is FlyRank's own observed
  search and engagement data, not access to how Google's systems work.
- I cannot treat a correlation as proof, especially a weak one (e.g. the
  -0.073 correlation between position and CTR) — a number that small could
  reflect noise, a non-linear relationship, or confounding factors I
  haven't controlled for.
- I cannot generalize findings from this ~30,000-row anonymized starter
  slice to the full ~79M-row warehouse without re-checking on that larger,
  more representative data.
